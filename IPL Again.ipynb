```python
#loading the required libraries
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
#loading the ipl matches dataset
data = pd.read_csv("IPLmatches.csv")
```

```python
#Looking at the number of rows and columns in the dataset
data.shape
```

```python
#Getting the frequency of most man of the match awards
man_of_the_match = data["player_of_match"].value_counts()
man_of_the_match
```

```python
#Getting the top 10 players with most man of the match awards
top10 = data["player_of_match"].value_counts().head(10)
top10
```

```python
#Getting the top 5 players with most man of the match awards
top5 = data["player_of_match"].value_counts().head()
top5
```

```python
#making a bar-plot for the top 5 players with most man of the match awards
plt.bar(top5.index, top5.values)
plt.xlabel("PLAYERS")
plt.ylabel("FREQUENCY")
plt.title("TOP-5 PLAYERS")
plt.show()
```

```python
# COUNT TOP-7 UMPIRES AND MAKE A BAR PLOT...
umpire = data["umpire1"].value_counts().head(7)
plt.bar(umpire.index, umpire.values)
plt.xlabel("Names")
plt.ylabel("Frequency")
plt.title("Top-7 Umpires")
plt.xticks(rotation=45)
plt.show()
```

```python
most_winning_team = data["winner"].value_counts()
plt.figure(figsize=(12,6))
plt.bar(most_winning_team.index, most_winning_team.values)
plt.xlabel("Teams")
plt.ylabel("Frequency", rotation=45)
plt.title("MOST WINNING TEAMS")
plt.xticks(rotation=90)
plt.yticks(range(0, most_winning_team.values.max()+10, 10), rotation= 45)
plt.ylim(0,120)
plt.show()
```

```python
# Getting the frequency of result column
freq_of_result = data["result"].value_counts().to_frame().reset_index(drop=True)  # Erasing last line
freq_of_result
```

```python
#Finding out the number of toss wins w.r.t each team
toss_win = data["toss_winner"].value_counts()
toss_win
```

```python
#Extracting the records where a team won batting first...
batting_first = data[data["win_by_runs"]>0]               # THIS IS THE WHOLE DATA OF ALL FIRST-BATTING WINNER TEAMS...
batting_first
```

```python
F_batting = data[data["win_by_runs"]>0]
team_freq = F_batting["winner"].value_counts()         # FREQUENCY OF ALL FIRST-BATTING WINNER TEAMS...
print(team_freq)
```

```python
#Looking at the head
after_fielding = data[data["win_by_runs"]>0]
count = after_fielding["winner"].value_counts().head()
print(count)
```

```python
#Making a bargraph...
plt.figure(figsize=(12,6))
plt.bar(count.index, count.values)
plt.xlabel("frequency")
plt.ylabel("Teams")
plt.title("TOP-5 TEAMS")
plt.xticks(rotation=45)
plt.yticks(range(0, count.values.max()+10,20),rotation=45)
plt.show()
```

```python
# making a histogram...
plt.figure(figsize=(12,6))
plt.hist(count.values, bins=10, color="lightblue",edgecolor="blue")     # THE GRAPH IS LOOKING WEIRED BCZ THE GIVEN DATA HAS LESS VARIATION...
plt.xlabel("WINNING MATCHES")
plt.ylabel("FREQUENCY")
plt.title("TOP-5 TEAMS")
plt.show()
```

```python
#Finding out the number of wins w.r.t each team after batting first...
first_batting = data[data["win_by_runs"]>0]
x = first_batting["winner"].value_counts()
print(x)
```

```python
# MAKING A HISTOGRAM...
plt.figure(figsize=(13,6))
plt.hist(x.values, bins=7, color="lightblue", edgecolor="blue")
plt.xlabel("FREQUENCY")
plt.ylabel("SCALE")
plt.title("WINNING TEAMS AFTER FIRST BATTING")
plt.yticks(rotation=45)
plt.show()
```

```python
#Making a bar-plot for top 3 teams with most wins after batting first
f_batting = data[data["win_by_runs"]>0]
y = f_batting["winner"].value_counts().head(3)
plt.figure(figsize=(12,6))
plt.bar(y.index, y.values)
plt.yticks(range(0, y.max()+20, 10), rotation=45)
plt.xticks(rotation=45)
plt.xlabel("TEAMS")
plt.ylabel("FREQUENCY")
plt.title("TOP-3 TEAMS")
plt.show()
```

```python
#Making a pie chart
plt.pie(y.values, autopct="%0.2f%%", labels = y.index)
plt.show()
```

```python
#extracting those records where a team has won after batting second
after_batting = data[data["win_by_wickets"]>0]
count = after_batting["winner"].value_counts()
print(count)
```

```python
#Making a pie chart for distribution of most wins after batting second
a_batting = data[data["win_by_wickets"]>0]
x = a_batting["winner"].value_counts().head(6)
plt.figure(figsize=(12,9))
plt.pie(x.values, autopct="%0.2f%%", labels=x.index)
plt.show()
```
